```python
class Area:
    def __init__(self,l,b):
        self.l=l
        self.b=b
    def area(self):
        a=self.l*self.b
        print(f"The area is = {a}")
class Volumn:
    def __init__(self,l,b,h):
        self.l=l
        self.b=b
        self.h=h
    def volumn(self):
        v=self.l*self.b*self.h
        print(f"The volumn is = {v}")
        
obj1=Area(10,5)
obj2=Volumn(10,5,2)

obj1.area()
obj2.volumn()
```

```python
class Parrot:
    def fly(self):
        print("Parrot can fly")
    def swim(self):
        print("Parrot can't swim")

class Penguin:
    def fly(self):
        print("Penguin can't fly")
    def swim(self):
        print("Penguin can swim")

#common interface
def flying_test(bird):
    bird.fly()

def swimming_test(birds):
    birds.swim()

blu=Parrot()
peggy=Penguin()

#passing the object
flying_test(blu)
flying_test(peggy)

print('\n')
swimming_test(blu)
swimming_test(peggy)
```

```python
ms={"Ram","Samira","Shyam","Raj","Hari","Sita"}
apple={"Samira","Raj","Arimas","Steave"}

print("Working on both company:",ms.intersection(apple))
print("Working on Merge company:",ms.union(apple))
```

```python
print("Working only in microfsoft",ms.difference(apple))
print("Working only in apple",apple.difference(ms))
```

```python
chem=int(input("Enter the marks of chemistry:"))
phy=int(input("Enter the marks of physic:"))
math=int(input("Enter the marks of math:"))
total=chem+phy+math
average=total/2
print("The total marks is:", total)
print("The average marks is:",average)

if total>90:
    print("Yayy! You got A+")
elif total>80:
    print("Yayy! You got A")
elif total>70:
    print("Yayy! You got B+")
elif total<40:
    print("Opps! You're fail")
else:
    print("Enter number")
```

```python
chem=int(input("Enter the marks of chemistry:"))
phy=int(input("Enter the marks of physic:"))
math=int(input("Enter the marks of math:"))
total=chem+phy+math
average=total/3

if total>=80:
    grade='a'
elif total>=70:
    grade='b'
elif total>=45:
    grade='c'
else:
    grade='f'

print("The total marks is:", total)
print("The average marks is:",average)
print("The grade is:",grade)
```

```python
#WAP to create a calculator
a=int(input("Enter the value of a:"))
b=int(input("Enter the value of b:"))
print("Please Enter operator")
o=input("+ or - or * or /")
if o == "+":
    print(a+b,"is the value")
elif o =="-":
    print("The value is", a-b)
elif o =="*":
    print(a*b)
elif o== "/" and b!=0:
    print(a/b)
elif o=="/" and b==0:
    print("The value of b can't be 0")
else:
    print("Invalid operator")
```

```python
bill=""
total=0
n=int(input("Enter the number of products: "))
for i in range(n):
    product=input("Enter the product name: ")
    price=int(input("Enter the price: "))
    qty=float(input("Enter the quantity: "))
    money=price*qty
    info=f"{product:<16} {price:<13} {qty:<13} {money:<13}"
    bill=bill+info+"\n"
    total=total+money
vat=0.13*total
all_total=vat+total
```

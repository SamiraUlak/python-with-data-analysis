```python
# login system
username='samira'
password='admin'
count=5
def login():
    global count
    try:
        if count<=0:
            print("You have reached limit")
            return
            
        user=input("Enter user=")
        pas =input("Enter password=")
        assert username==user and password==pas
    except AssertionError:
        print("Incorrect username and password",count)
        count-=1
        login()
    else:
        print("Welcome")
login()
print(f'You have loged in {count} times')
    
```

```python
#checking withdrawal amount with balance
balance=10000
count=1
def check():
    global balance
    global count
    try:
        while True:
            choice=int(input("\nEnter 1 to withdraw ,0 to stop"))
            if choice==1:
                amount=int(input("Enter the amount you want to withdraw="))
                if amount<=balance:
                    balance=balance-amount
                    print(f'Transation Successful.\nYou have Rs.{balance} in your account')
                    count+=1  
                else:
                    print("Insufficient Balance")
                    break     
            if choice==0:
                print(f"Thanks! You have Rs.{balance} in your account")
                break
    except ValueError:
        print("Error! Please enter interger number.")     
check()
print(f'You have withdrawn {count} times')
         
    
```

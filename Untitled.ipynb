```python
#insert 20 random letters range a through f
import random

letters=['a','b','c','d','e','f']
result=[]
for i in range(20):
    a=random.choice(letters)
    result.append(a)
print(result)  
```

```python
#sort the list in ascending order
result.sort()
print(result)

print("ascending order",sorted(result))
```

```python
#sort in descending order
result.reverse()
print(result)

print("descending order",sorted(result,reverse=True))

result.sort(reverse=True)
print(result)
```

```python
unique=set(result)
unique=list(unique)
print(sorted(unique))
```

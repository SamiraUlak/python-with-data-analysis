```python
class Student:
    #default constructor
    def __init__(self):
        pass

    #parameterized constructor
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
        
    #methods
    def welcome(self):
        print("welcome Students",self.name)

    def get_marks(self):
        return self.marks
        
        
s1=Student("Sami",97)
print(s1.name,s1.marks)

s2=Student("Raj",98)
print(s2.name,s2.marks)

s1.welcome()
s1.get_marks()
```

```python
class Student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks

    def get_avg(self):
        sum=0
        for i in self.marks:
            sum+=i
        avg=sum/3
        print("hi",self.name,"your average score is:",avg)


s1=Student("Samira",[98,68,99])
s1.get_avg()
            
```

```python
class Student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks

    def get_avg(self):
        sum=0
        for i in range(len(self.marks)):
            sum=sum+self.marks[i]
        avg=sum/3
        print("hi",self.name,"your average score is:",avg)


s1=Student("Samira",[98,68,99])
s1.get_avg()
            
```

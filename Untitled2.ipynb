```python
l=[]
i=1
while i<=10:
    j=i**2
    l.append(j)
    i+=1
print(l)

t=tuple(l)
print(t)

x=int(input("Enter the value of x="))
i=0
while i<len(t):
    if t[i]==x:
        print("Congratss!! You're Lucky")
        break
    i+=1
else:
    print("opps")

```

```python
# class <class_name>:            #class
#     <methods>                  #methods
#     <methods>
# obj=<class_name>()             #object
```

```python
class Hello:
    
    def hello(self):
        print("Hello World")
obj=Hello()
obj.hello()
```

```python
class Hello:
    @staticmethod
    def hello():
        print("Hello World")
obj=Hello()
obj.hello()
Hello.hello()
```

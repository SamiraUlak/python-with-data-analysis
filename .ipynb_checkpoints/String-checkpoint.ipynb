```python
bill=""
all_total=0
n=int(input("Enter the value n="))
for i in range(n):
    name=input("Product name")
    price=int(input("Enter price"))
    qty=float(input("Enter qty"))
    total=price*qty
    info=f"{name} {price} {qty} {total}\n"
    bill=bill+info
    all_total=all_total+total
print("name price quantity total")
print(bill)
print("Your total\n",all_total)
```

```python
bill=""
total=0
n=int(input("Enter the number of products: "))
for i in range(n):
    product=input("Enter the product name: ")
    price=int(input("Enter the price: "))
    qty=float(input("Enter the quantity: "))
    money=price*qty
    info=f"{product:<16} {price:<13} {qty:<13} {money:<13}"
    bill=bill+info+"\n"
    total=total+money
vat=0.13*total
all_total=vat+total
```

```python
print("="*50)
print(f"{'Welcome':^50}")
print("="*50)
print(f"{'product':<16} {'price':<13} {'Quantity':<13} {'money':<13}")
print("-"*50)
print(bill)
print("-"*50)
print(f"{'Total':<44} {total}")
print(f"{'Vat(13%)':<44} {vat}")
print("-"*50)
print(f"{'All total':<44} {all_total}")
print("="*50)
```

```python
name= input("Enter the name of student:")
roll=int(input("Enter your roll:"))
grade=int(input("Enter the grade:"))
chem=int(input("Enter the chem marks:"))
phy=int(input("Enter the phy marks:"))
math=int(input("Enter the math marks:"))
subject={"Chemistry":chem,"Physic":phy,"Math":math}
total=chem+phy+math
average=total/3
cgpa=average/25
if cgpa>=3.6:
    Grade="A+"
elif cgpa>=3.2:
    Grade="A"
elif cgpa>=2.8:
    Grade="B+"
elif cgpa>=2.4:
    Grade="B"
elif cgpa>=2.0:
    Grade="C+"
elif cgpa>=1.6:
    Grade="C"
else:
    Grade="F"
```

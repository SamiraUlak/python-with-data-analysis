```python
n = int(input("Enter the number of student: "))
result = ""

for i in range(n):
    name = input("Enter the name of student: ")
    roll = int(input("Enter your roll: "))
    grade = int(input("Enter the grade: "))
    chem = int(input("Enter the chem marks: "))
    phy = int(input("Enter the phy marks: "))
    math = int(input("Enter the math marks: "))

    total = chem + phy + math
    average = total / 3
    cgpa = average / 25

    if chem >= 45 and phy >= 45 and math >= 45:
        if average >= 90:
            Grade = "A+"
        elif average >= 80:
            Grade = "A"
        elif average >= 70:
            Grade = "B+"
        elif average >= 60:
            Grade = "B"
        elif average >= 50:
            Grade = "C+"
        elif average > 45:
            Grade = "C"
    else:
        Grade = "F"

    result += "="*50 + "\n"
    result += f"{'Mount Valley English School':^50}\n"
    result += f"{'Kwathandou,Bhaktapur':^50}\n"
    result += "="*50 + "\n"
    result += f"Name={name}{' '*9}Roll:no={roll}{' '*9}Grade={grade}\n"
    result += "-"*50 + "\n"
    result += f"{'S.N':<9}{'Subject':<34}{'Marks':<19}\n"
    result += f"{'1':<9}{'Chemistry':<34}{chem}\n"
    result += f"{'2':<9}{'Physic':<34}{phy}\n"
    result += f"{'3':<9}{'Math':<34}{math}\n"
    result += "-"*50 + "\n"
    result += f"{'Total':<43}{total}\n"
    result += "-"*50 + "\n"
    result += f"Grade:{Grade:<30}Cgpa:{cgpa:.2f}\n\n"

print(result)

file=open("Data2.txt",'w')
file.write("Name Price Quantity, total")
file.write(result)
file.close()
```

```python
bill=""
all_total=0
n=int(input("Enter the value n="))
for i in range(n):
    name=input("Product name")
    price=int(input("Enter price"))
    qty=float(input("Enter qty"))
    total=price*qty
    info=f"{name} {price} {qty} {total}\n"
    bill=bill+info
    all_total=all_total+total
print("name price quantity total")
print(bill)
print("Your total\n",all_total)

file= open("Bills.txt",'w')
file.write(bill)
file.write(f'{all_total}')
file.close()
```

```python
import csv
csv_columns=["name","price",'qty','total']
dict=[
    {'name': 'momo', 'price': '200', 'qty': '3.0', 'total': '600.0'},
    {'name': 'coke', 'price': '100', 'qty': '3.0', 'total': '300.0'}
]
file=open("Dict_data.csv",'w')
writer=csv.DictWriter(file,fieldnames=csv_columns)
writer.writeheader()
for data in dict:
    writer.writerow(data)
```

```python
#function with argument and return type
def add(a,b):
    return a+b
def sub(a,b):
    return a-b
def mul(a,b):
    return a*b
def div(a,b):
    return a/b
    
a=int(input("Enter a="))
b=int(input("Enter b="))
o=input("Enter o=")
if o=='+':
    print(add(a,b))
elif o=='-':
    print(sub(a,b))
elif o=='*':
    print(mul(a,b))
elif o=="/":
    print(div(a,b))
else:
    print("Please choose any operator")
    
```

```python
#function with argument and return type
def area(l,b):   #parameter
    return l*b
def volumn(l,b,h):
    return area(l,b)*h
    
l=int(input("Enter l="))
b=int(input("Enter b="))

ar=area(l,b)  #argument
print("The area =",ar)

h=int(input("Enter h="))
vr=volumn(l,b,h)
print("The volumn =",vr)
```

```python
infos=""
def info(name,phone):
    global infos
    a=f"{name}---{phone}\n"
    infos=infos + a
    
n=int(input("Enter n="))
for i in range(n):
    name=input("Enter name=")
    phone=input("Enter phone=")
    info(name,phone)
```

```python
# lambda function------function having no name
#  # x=lambda <argument>:<operation>
# x(<argument>)
```

```python
x=lambda a:a**2
print(x(5))
```

```python
x=lambda l,b:l*b
print(x(2,3))
```

```python
#filter(),map()
#list(filter(lambda <argument>:<condition>,list))

l=[21,2,33,43,54,32,56,77,88,11,23]
list(filter(lambda x:x%2==0,l))
```

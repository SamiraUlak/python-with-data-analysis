```python
# Multi level inheritance
class MyAddress:
    def __init__(self,add):
        self.add=add

    def adds(self):
        print(self.add)
        
class Data(MyAddress):
    def __init__(self,name,age,add):
        self.name=name
        self.age=age
        MyAddress.__init__(self,add)
        
    def data(self):
        print(self.name,self.age)
        
class Info(Data):
    def __init__(self,name,age,add,phone):
        self.phone=phone
        Data.__init__(self,name,age,add)
        
    def info(self):
        print(f"My name is {self.name}. I am from {self.add}. I am {self.age} years old. My phone is {self.phone} ")

name=input("Enter name=")
age=int(input("Enter age="))
add=input("Enter address=")
phone=input("Enter phone no=")

obj=Info(name,age,add,phone)
obj.info()
obj.data()
obj.adds()
```

```python
#Multiple Inheritance
class MyAddress:
    def __init__(self,add):
        self.add=add
    def adds(self):
        print(self.add)

class Data:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def infos(self):
        print(self.name,self.age)

class Info(Data,MyAddress):
    def __init__(self,add,name,age,phone):
        self.phone=phone
        MyAddress.__init__(self,add)
        Data.__init__(self,name,age)
        
    def info(self):
        print(f"My name is {self.name}. I am from {self.add}. I am {self.age} years old. My phone is {self.phone} ")

name=input("Enter name=")
age=int(input("Enter age="))
add=input("Enter add=")
phone=input("Enter phone=")
obj=Info(add,name,age,phone)
obj.info()
obj.infos()
obj.adds()
```

```python
## accessing private member to other class
class Data:
    def __init__(self,name,age,add):
        self.name=name
        self._age=age
        self.__add=add
        
    def data(self):
        print(self.name,self._age,self.__add)

    def myadd(self):
        return self.__add

#Child Class
class Info(Data):
    def info(self):
        ad=Data.myadd(self)
        print(ad)
        a=f"Hello World I am {self.name}. I am{self._age}."
        print(a)

name=input("Enter name=")
age=int(input("Enter age="))
add=input("Enter address=")

obj=Info(name,age,add)
obj.info()
obj.data()
```
